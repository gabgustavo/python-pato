```python
import pandas as pd
```

```python
# para este ejercicio se esta asumiendo que la información ya esta limpia
# para hacer una limpiesa de la infromacion se debe:
# 1. limpiar los datos nulos
# 2. depurar la información duplicada
# 3. depurar la información incorrecta
# 4. limpiar la información no relevante
data = pd.read_csv('data/juegos-ml.csv')
data
```

```python
# Datos de entrada, es la información que se le pasa al modelo para que nos de un resultado
# Los datos de dalida es la informacion que esperamos que el modelo nos de como resultado, segun la entrada

# X es por convención la variable que contiene los datos de entrada, es decir, la información que se le pasa al modelo para que nos de un resultado
X = data.drop(columns=['juegos'])
X
```

```python
y = data['juegos']
y
```

```python
# https://scikit-learn.org/stable/user_guide.html
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X.values, y)
predicciones = model.predict([[14, 0]])
predicciones
```

```python
predicciones = model.predict([[14, 0], [29, 0], [30, 0]])
predicciones
```
