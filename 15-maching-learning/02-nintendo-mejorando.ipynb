```python
import pandas as pd
# https://scikit-learn.org/stable/user_guide.html
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# para este ejercicio se esta asumiendo que la información ya esta limpia
# para hacer una limpiesa de la infromacion se debe:
# 1. limpiar los datos nulos
# 2. depurar la información duplicada
# 3. depurar la información incorrecta
# 4. limpiar la información no relevante
data = pd.read_csv('data/juegos-ml.csv')
# data
X = data.drop(columns=['juegos'])
#X
y = data['juegos']
#y

X_entrenar, X_prueba, y_entrenar, y_prueba = train_test_split(X.values, y, test_size=0.2)

model = DecisionTreeClassifier()
#model.fit(X.values, y)
model.fit(X_entrenar, y_entrenar)
#predicciones = model.predict([[14, 0]])
#predicciones = model.predict([[14, 0], [29, 0], [30, 0]])
predicciones = model.predict(X_prueba)
predicciones
```

```python
# ahora ya que se tiene los datos de entrenamiento separados de los de prueba, ahora debemos comparar las predicciones que nos entrego el modelo con los datos de prueba
puntaje = accuracy_score(y_prueba, predicciones)
puntaje
```
